```python
import pandas as pd

df= pd.read_csv("sports.csv")
df.head() #näitab first rows #df näitab kõiki ridu
```

```python
df.describe() # teeb basic statistikat iga arvulise veeru jaoks
```

```python
#Box plot to each athlete to find the number of years of practise 
```

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Years of practice")
plt.ylabel("Years")
plt.title("Years of practice per athlete")
plt.show
```

```python
df.plot(kind="bar", x="Name", y="Age")

plt.show
```

```python
plt.scatter(df["Age"],df["Years of practice"], s=df["Years of practice"]*30, alpha=0.6)
plt.show    #Takes x and y                     #Teeb pallid suuerams,      #Teeb läbipaistvamaks
```

```python
#Average age per sport
```

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.show
```
